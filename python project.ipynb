```python

import pyttsx3
import speech_recognition as sr
import webbrowser
import shutil
import pywhatkit
import datetime
import pandas as pd
```

```python
engine=pyttsx3.init()
engine.setProperty("rate",120)
voice=engine.getProperty("voices")
engine.setProperty('voice',voice[0].id)

def speak(command:str):
     engine.say(command)
     engine.runAndWait()
     
     
def voice_recognizing():
    content=" "
    while content == " ":
        r = sr.Recognizer()
        with sr.Microphone() as source:
            print("Say something!")
            audio = r.listen(source)
        try:
            content= r.recognize_google(audio,language='en-ind')
            print("you said..."+content)
        except sr.UnknownValueError:
            print("could not understand audio")
        except sr.RequestError as e:
            print("Could not request results  {0}".format(e))
        except Exception as e:
            print(f"the exception occured is {e}")
               
    return content

def main():
    while True:
            voicereco=voice_recognizing().lower()
            speak("welcome Sir I am ur Assistant")
            try:
                if "hello" in voicereco:
                    speak("how can i help u sir")
                elif "how are you"in voicereco:
                    speak("i am good sir ")
                elif "play music" in voicereco:
                     speak("which music do u want me to play sir")
                     while True:
                            try:
                                song_name={"Blinding Lights": 'The Weeknd',
                                        "Shape of You":' Ed Sheeran',
                                        "Levitating": 'Dua Lipa (feat. DaBaby)',
                                        "Believer" : 'Imagine Dragons',
                                        "Dance Monkey" : 'Tones and I'}
                                print("song and artist names availabe in the system")
                                for index,(key,value) in enumerate(song_name.items()):
                                    print(f"{index + 1}. Song: {key} by {value}")
                                speak("which song would u like to play sir")
                                a=voice_recognizing().lower()
                                if a in song_name:
                                    play_song(a,song_name[a])
                                elif a not in song_name:
                                    speak('which song do u want to add')
                                    result = voice_recognizing().lower()
                                    song_name[result]='unkown artis'
                                    speak(f"The song {result} has been added to the system.")
                                    continue
                                elif a is "stop playing":
                                    break
                            
                            except Exception as e:
                                print(f"the error occure is {e}")
                elif 'current time' in voicereco:
                     current_time=datetime.datetime.now().strftime("%H:%M")
                     speak(" SIR the current time is "+ str(current_time))
                elif 'todays date' in voicereco:
                    today_date=datetime.datetime.now().strftime("%d:%m")
                    speak("SIR todays date is   "+ str(today_date))
                    break
                elif 'make a to do list','new task':
                    
                             
                             
                             
                        
                             
                                                               
                               
                                
                                
                    
 
                    
 def play_song(song:str,artist:str):
     try:
         query=f"{song} {artist}"
         pywhatkit.playonyt(query)
         print(f"playing{song} by {artist}" )
     except Exception as e:
         print(f'song error {e}')        
        
                
    
if __name__=='__main__':
    main()
    
```
