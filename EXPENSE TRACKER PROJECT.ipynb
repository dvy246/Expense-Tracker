```python
# question 1
#flatten a nest list 

def flattend_list(lst):
    new=[]
    for el in lst:
        if isinstance(el,list):
            new.extend(flattend_list(el))
            
        else:
            new.append(el)
        
    return new
    
    
flattend_list([1, [2, [3, 4], 5], 6])
```

```python
def merge(left,right):
    new_l=[]
    if len(left)!=len(right):
        print("cant join")
    for i in range(len(left)):
         new_l.append(left[i])
         new_l.append(right[i])
    return new_l
        
merge([1,2,3,4],[2,3,4,5])
```
